```sql
SELECT
    delay_category,
    COUNT(*) AS stop_visits
FROM (

    SELECT 
        CASE 
            WHEN delay_minutes < -1 THEN '1. Early (1+ min)'
            WHEN delay_minutes BETWEEN -1 AND 5 THEN '2. On time (1 early-5 late)'
            WHEN delay_minutes > 5 AND delay_minutes <= 10 THEN '3. Late (5-10 min)'
            WHEN delay_minutes > 10 AND delay_minutes <= 20 THEN '4. Very late (10-20 min)'
            ELSE '5. Severely late (20+ min)'
        END AS delay_category

    FROM bc_transit_ws.gold.stop_level_on_time

    WHERE delay_minutes IS NOT NULL

) t

GROUP BY delay_category

ORDER BY delay_category;
```

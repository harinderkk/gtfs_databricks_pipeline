```sql
WITH popular_routes AS (
    SELECT route_name, route_rank
    FROM (VALUES
        ('099', 1),
        ('R4',  2),
        ('049', 3),
        ('025', 4),
        ('016', 5),
        ('R6',  6),
        ('R1',  7),
        ('323', 8),
        ('019', 9),
        ('R5',  10)
    ) AS t(route_name, route_rank)
)
SELECT 
    CONCAT(p.route_name, ' - ', t.route_long_name) AS route,
    CASE t.fetch_window 
        WHEN '7am' THEN 'A. Morning (7am)'
        WHEN '12pm' THEN 'B. Midday (12pm)'
        WHEN '5pm' THEN 'C. Evening (5pm)'
    END AS time_window,
    t.pct_on_time
FROM bc_transit_ws.gold.route_time_of_day_performance t
JOIN popular_routes p ON t.route_name = p.route_name
WHERE t.fetch_window IN ('7am', '12pm', '5pm')
ORDER BY p.route_rank, time_window
```

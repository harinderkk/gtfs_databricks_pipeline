```sql
SELECT 
    CONCAT(route_name, ' - ', route_long_name) AS route,
    avg_scheduled_headway_min AS scheduled_wait,
    avg_actual_headway_min AS actual_wait,
    ROUND(avg_actual_headway_min - avg_scheduled_headway_min, 2) AS slip_min
FROM bc_transit_ws.gold.route_passenger_wait_times
WHERE total_gaps >= 1000
ORDER BY avg_actual_headway_min ASC
LIMIT 15
```

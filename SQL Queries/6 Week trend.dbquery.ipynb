```sql
SELECT 
    obs_date,
    ROUND(SUM(is_on_time) * 100.0 / COUNT(*), 2) AS daily_on_time_pct,
    ROUND(AVG(ABS(delay_minutes)), 2) AS daily_avg_deviation
FROM bc_transit_ws.gold.stop_level_on_time
WHERE delay_minutes IS NOT NULL
AND ABS(delay_minutes) <= 60
AND HOUR(actual_arrival) IN (7, 12, 17)
GROUP BY obs_date
ORDER BY obs_date
```

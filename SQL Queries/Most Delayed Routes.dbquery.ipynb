```sql
WITH routes_with_all_windows AS (
    SELECT route_name
    FROM bc_transit_ws.gold.route_time_of_day_performance
    WHERE fetch_window IN ('7am', '12pm', '5pm')
    GROUP BY route_name
    HAVING COUNT(DISTINCT fetch_window) = 3
)
SELECT 
    CONCAT(r.route_name, ' - ', r.route_long_name) AS route,
    r.avg_delay_minutes AS avg_deviation_min,
    r.on_time_pct
FROM bc_transit_ws.gold.route_on_time_performance r
JOIN routes_with_all_windows w ON r.route_name = w.route_name
WHERE r.total_stop_observations >= 1000
ORDER BY r.avg_delay_minutes DESC
LIMIT 10
```

```sql
WITH routes_with_all_windows AS (
    SELECT route_name
    FROM bc_transit_ws.gold.route_time_of_day_performance
    WHERE fetch_window IN ('7am', '12pm', '5pm')
    GROUP BY route_name
    HAVING COUNT(DISTINCT fetch_window) = 3
),
top_routes AS (
    SELECT r.route_name, r.avg_delay_minutes AS overall_deviation
    FROM bc_transit_ws.gold.route_on_time_performance r
    JOIN routes_with_all_windows w ON r.route_name = w.route_name
    WHERE r.total_stop_observations >= 1000
    ORDER BY r.avg_delay_minutes ASC
    LIMIT 5
)
SELECT 
    CONCAT(t.route_name, ' - ', t.route_long_name) AS route,
    CASE t.fetch_window 
        WHEN '7am'  THEN 'A. Morning (7am)'
        WHEN '12pm' THEN 'B. Midday (12pm)'
        WHEN '5pm'  THEN 'C. Evening (5pm)'
    END AS fetch_window,
    t.pct_on_time,
    tr.overall_deviation
FROM bc_transit_ws.gold.route_time_of_day_performance t
JOIN top_routes tr ON t.route_name = tr.route_name
WHERE t.fetch_window IN ('7am', '12pm', '5pm')
ORDER BY tr.overall_deviation ASC, t.route_name, fetch_window
```

```sql
SELECT
DATE_FORMAT (obs_date, 'EEEE') AS day_of_week,
DAYOFWEEK (obs_date) AS dow_num,
ROUND (SUM (is_on_time) * 100.0 / COUNT (*), 2) AS on_time_pct,
ROUND (AVG (ABS (delay_minutes)), 2) AS avg_deviation
FROM bc_transit_ws.gold.stop_level_on_time
WHERE delay_minutes IS NOT NULL
GROUP BY DATE_FORMAT (obs_date, 'EEEE'), DAYOFWEEK(obs_date)
ORDER BY dow_num;
```

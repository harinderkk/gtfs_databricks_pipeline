```python

CATALOG_NAME = "bc_transit_ws" 
ANALYTICS_SCHEMA = "transit_analytics" 
print(f"The new schema for : {CATALOG_NAME}.{ANALYTICS_SCHEMA}")

# --- 2. Create the Schema (if it doesn't exist) ---
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG_NAME}.{ANALYTICS_SCHEMA}")

# --- 3. Set the Session Context (CRITICAL STEP) ---
# This tells Spark: "When I say SELECT * FROM table_a, look in transit_analytics."
spark.sql(f"USE CATALOG {CATALOG_NAME}")
spark.sql(f"USE SCHEMA {ANALYTICS_SCHEMA}")

print("Unity Catalog session context is set.")
```
